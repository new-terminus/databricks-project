# Permisos de lectura de Gold

Se concede acceso de lectura al grupo `bitcoin_readers` sobre las dos tablas de Gold.


```python
%sql
GRANT USE CATALOG ON CATALOG catalog_smartdata TO `bitcoin_readers`;
GRANT USE SCHEMA ON SCHEMA catalog_smartdata.bitcoin_gold TO `bitcoin_readers`;
```

```python
%sql
GRANT SELECT ON TABLE catalog_smartdata.bitcoin_gold.bitcoin_daily_metrics TO `bitcoin_readers`;
GRANT SELECT ON TABLE catalog_smartdata.bitcoin_gold.sentiment_summary TO `bitcoin_readers`;
```
